# **Project Summary -**

The Mental Health in Tech Survey Analysis project is an Exploratory Data Analysis (EDA) project based on survey data related to mental health attitudes, experiences, and workplace support within the technology sector. The main objective of this project is to understand patterns in mental health treatment, workplace attitudes, employee awareness, and the availability of mental health support provided by employers.

The dataset originally contained 1,259 rows and 27 columns. It included demographic information such as age, gender, country, and state, along with employment-related and mental health variables such as self-employment, family history of mental illness, treatment, work interference, remote work, employer benefits, care options, wellness programs, help resources, anonymity, medical leave, workplace consequences, and willingness to discuss mental health with coworkers or supervisors.

The project began with understanding the structure and characteristics of the dataset. The data was inspected for its shape, data types, unique values, duplicate records, and missing values. Missing values were identified in the state, self_employed, work_interfere, and comments columns. The Age column was also examined and unrealistic values, including negative and extremely large ages, were identified. To prepare the dataset for analysis, respondents with unrealistic ages were removed and ages between 18 and 100 were retained. Missing state values were replaced with “Not Applicable,” missing self_employed and work_interfere responses were labelled “Unknown,” and missing comments were replaced with “No Comment.” No duplicate rows were found, and the cleaned dataset contained no remaining missing values.

Multiple visualizations were then created using Python, Pandas, Matplotlib, and Seaborn to explore important patterns in the survey. The analysis showed that respondents who reported a family history of mental illness were more commonly represented among those who had sought mental health treatment. Treatment-seeking was also more common among respondents who reported that mental health sometimes or often interfered with their work.

Workplace-related analysis revealed several areas of uncertainty. Many respondents did not know whether their employer protected anonymity when using mental health resources or whether mental health was treated as seriously as physical health. A large group was also uncertain about the ease of taking medical leave for a mental health condition. Awareness of employer-provided mental health care options and help resources was inconsistent.

The analysis also showed that many respondents reported that mental health was not discussed as part of their employer’s wellness program. Respondents appeared more hesitant to discuss a mental health condition during a job interview compared with a physical health condition. However, most respondents reported that they had not observed negative consequences for coworkers with mental health conditions.

Overall, the analysis highlights the importance of improving mental health awareness, communication, confidentiality, and accessibility of workplace support. Organizations can benefit from clearly communicating available mental health benefits, care options, leave policies, and confidential resources. They can also strengthen wellness initiatives and provide appropriate awareness training for managers and supervisors. These measures can contribute to a workplace environment where employees are better informed about available mental health support and feel more comfortable accessing appropriate resources when needed.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
df_clean=df.copy()
````

````python
df_clean['Age'].sort_values().unique()
````

````python
df_clean = df_clean[(df_clean['Age'] >= 18) & (df_clean['Age'] <= 100)]
````

````python
df_clean['Age'].sort_values().unique()
````

````python
df_clean['state'] = df_clean['state'].fillna('Not Applicable')
````

````python
df_clean['self_employed'] = df_clean['self_employed'].fillna('Unknown')
````

````python
df_clean['work_interfere'] = df_clean['work_interfere'].fillna('Unknown')
````

````python
df_clean['comments'] = df_clean['comments'].fillna('No Comment')
````

````python
df_clean.isnull().sum()[df_clean.isnull().sum() > 0]
````

### What all manipulations have you done and insights you found?

The following data wrangling steps were performed to make the dataset ready for analysis:

1. Created a copy of the original dataset to preserve the raw data.
2. Checked the Age column and found unrealistic values such as negative ages and extremely large values.
3. Removed records with unrealistic ages and retained respondents between 18 and 100 years. After cleaning, the observed age range is 18 to 72 years.
4. Missing values in the state column were replaced with "Not Applicable" because state information is relevant only to respondents from the United States.
5. Missing values in self_employed and work_interfere were replaced with "Unknown" to avoid making assumptions about unanswered survey questions.
6. Missing values in comments were replaced with "No Comment" because this was an optional comment field.
7. The dataset contained no duplicate rows.
8. After data wrangling, the dataset was checked again and no missing values remained.

The cleaned dataset is now ready for exploratory data analysis and visualization.

#### Chart - 14 - Correlation Heatmap

````python
# Select relevant categorical variables and convert them into numerical values
heatmap_data = df_clean[
    ['treatment', 'family_history', 'remote_work', 'tech_company']
].copy()

# Convert Yes/No values into 1/0
for column in heatmap_data.columns:
    heatmap_data[column] = heatmap_data[column].map({'Yes': 1, 'No': 0})

# Calculate the correlation matrix
correlation_matrix = heatmap_data.corr()

# Create correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of Selected Mental Health Variables')
plt.tight_layout()
plt.show()
````

### 1. Why did you pick the specific chart?

I selected a correlation heatmap because it provides a simple visual representation of the relationships between multiple numerical variables. The categorical Yes/No variables were converted into numerical values so that their correlations could be examined.

### 2. What is/are the insight(s) found from the chart?

The heatmap shows the correlations among treatment, family history, remote work, and whether the respondent works in a tech company. Most of the relationships are relatively weak, indicating that these variables do not have strong linear relationships with each other. Family history shows some relationship with treatment, while remote work and tech company status show weaker relationships with treatment.

### 3. Will the gained insights help creating a positive business impact? Are there any insights that lead to negative growth? Justify with specific reason.

Yes, the heatmap helps organizations understand that mental health treatment behaviour cannot be explained strongly by a single factor among the variables included in this analysis. Employers should therefore consider multiple personal and workplace factors when developing mental health support initiatives. The correlations show associations only and should not be interpreted as evidence that one variable directly causes another.
